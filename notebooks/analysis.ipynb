# Klasyfikacja wiadomości spam SMS

Celem ćwiczenia było porównanie różnych metod reprezentacji tekstu
oraz klasyfikatorów w zadaniu wykrywania wiadomości spam.

Wykorzystano zbiór SMS Spam Collection zawierający 5572 wiadomości.

# Preprocessing tekstu

Wykonano następujące operacje:

- zamiana tekstu na małe litery,
- usunięcie adresów URL,
- usunięcie znaków specjalnych,
- tokenizacja,
- usunięcie stopwords,
- lematyzacja.
  
# Reprezentacje tekstu

Badane reprezentacje:

1. Bag of Words (CountVectorizer)
2. TF-IDF (unigramy oraz bigramy)
3. Word2Vec + średnia wektorów słów

# Klasyfikatory

W eksperymencie wykorzystano:

1. Logistic Regression
2. Multinomial Naive Bayes
3. Linear Support Vector Machine (LinearSVM)
